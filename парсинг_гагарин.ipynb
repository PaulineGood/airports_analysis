```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загрузка данных из файла
with open('all_flights.json', 'r', encoding='utf-8') as f:
    flights = json.load(f)

# Преобразуем данные в DataFrame
df = pd.DataFrame(flights)

# Преобразуем время отправления и прибытия в числовой формат (часы)
df['Отправление_час'] = df['Отправление'].str.split(':').str[0].astype(int)
df['Прибытие_час'] = df['Прибытие'].str.split(':').str[0].astype(int)

# 1. График: Распределение рейсов по авиакомпаниям
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='Авиакомпания', order=df['Авиакомпания'].value_counts().index)
plt.title('Распределение рейсов по авиакомпаниям')
plt.xlabel('Количество рейсов')
plt.ylabel('Авиакомпания')
plt.show()
```

```python
# 2. График: Распределение рейсов по дням недели
# Создаем список всех дней недели
days_of_week = ['пн', 'вт', 'ср', 'чт', 'пт', 'сб', 'вс']

# Функция для подсчета количества рейсов по дням недели
def count_flights_by_day(df):
    day_counts = {day: 0 for day in days_of_week}  # Инициализируем счетчик для каждого дня
    for days in df['Дни полетов']:
        for day in days.split(', '):  # Разделяем строку на отдельные дни
            if day in day_counts:
                day_counts[day] += 1
    return day_counts

# Получаем данные для графика
day_counts = count_flights_by_day(df)

# Преобразуем данные в DataFrame для удобства построения графика
day_counts_df = pd.DataFrame(list(day_counts.items()), columns=['День недели', 'Количество рейсов'])

# Упорядочиваем дни недели
day_counts_df['День недели'] = pd.Categorical(day_counts_df['День недели'], categories=days_of_week, ordered=True)
day_counts_df = day_counts_df.sort_values('День недели')

# Построение графика
plt.figure(figsize=(10, 6))
sns.barplot(data=day_counts_df, x='День недели', y='Количество рейсов', palette='viridis')
plt.title('Распределение рейсов по дням недели')
plt.xlabel('День недели')
plt.ylabel('Количество рейсов')
plt.show()
```

```python
import time
import requests
from bs4 import BeautifulSoup
import json
import re
from datetime import datetime

# Настройки для обхода ограничений
HEADERS = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.5'
}

# Функция для безопасного парсинга с повторными попытками
def safe_parse(url, max_retries=3):
    for attempt in range(max_retries):
        try:
            response = requests.get(url, headers=HEADERS, timeout=10)
            response.raise_for_status()
            return BeautifulSoup(response.text, 'html.parser')
        except Exception as e:
            print(f"Попытка {attempt + 1} не удалась: {str(e)[:100]}...")
            time.sleep(10 * (attempt + 1))  # Увеличиваем задержку с каждой попыткой
    return None

def parse_schedule(url):
    try:
        soup = safe_parse(url)
        if not soup:
            return None, []

        # Извлекаем дату из URL Wayback Machine
        date_match = re.search(r'/web/(\d{4})(\d{2})(\d{2})', url)
        if date_match:
            capture_date = f"{date_match.group(1)}-{date_match.group(2)}-{date_match.group(3)}"
        else:
            capture_date = datetime.now().strftime("%Y-%m-%d")  # Для текущего расписания

        table = soup.find('table')
        if not table:
            print(f"Таблица не найдена на странице: {url}")
            return capture_date, []

        flights = []
        rows = table.find_all('tr')
        for row in rows[1:]:  # Пропускаем заголовок
            cols = row.find_all('td')
            if len(cols) >= 8:  # Проверяем, что строка содержит данные
                flight = {
                    "capture_date": capture_date,
                    "departure_time": cols[0].text.strip(),
                    "arrival_time": cols[1].text.strip(),
                    "flight_days": cols[2].text.strip(),
                    "period": cols[3].text.strip(),
                    "airline": cols[4].text.strip(),
                    "flight_number": cols[5].text.strip(),
                    "aircraft_type": cols[6].text.strip(),
                    "destination": cols[7].text.strip(),
                    "source_url": url
                }
                flights.append(flight)
        return capture_date, flights

    except Exception as e:
        print(f"Критическая ошибка при обработке {url}: {str(e)[:100]}...")
        return None, []

# Список URL-адресов для парсинга
urls = [ "https://web.archive.org/web/20190816013535/https://gsv.aero/schedule/", "https://web.archive.org/web/20190819090645/https://gsv.aero/schedule/", "https://web.archive.org/web/20190827183142/https://gsv.aero/schedule/", "https://web.archive.org/web/20200716105639/https://gsv.aero/schedule/", "https://web.archive.org/web/20200813172321/https://gsv.aero/schedule/", "https://web.archive.org/web/20200907230457/https://gsv.aero/schedule/", "https://web.archive.org/web/20200926021205/https://gsv.aero/schedule/", "https://web.archive.org/web/20201130142132/https://gsv.aero/schedule/", "https://web.archive.org/web/20210123013501/https://gsv.aero/schedule/", "https://web.archive.org/web/20210130204528/https://gsv.aero/schedule/", "https://web.archive.org/web/20210324080635/https://gsv.aero/schedule/", "https://web.archive.org/web/20210325051725/https://gsv.aero/schedule/", "https://web.archive.org/web/20210414140914/https://gsv.aero/schedule/", "https://web.archive.org/web/20210621041555/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210621041555/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210801233520/https://gsv.aero/schedule/", "https://web.archive.org/web/20210909135810/https://gsv.aero/schedule/", "https://web.archive.org/web/20210929020818/https://gsv.aero/schedule/", "https://web.archive.org/web/20211028175255/https://gsv.aero/schedule/", "https://web.archive.org/web/20220123060951/https://gsv.aero/schedule/", "https://web.archive.org/web/20220128073221/https://gsv.aero/schedule/", "https://web.archive.org/web/20220319023258/https://gsv.aero/schedule/", "https://web.archive.org/web/20220515100822/https://gsv.aero/schedule/", "https://web.archive.org/web/20220526210722/https://gsv.aero/schedule/", "https://web.archive.org/web/20221006204547/https://gsv.aero/schedule/", "https://web.archive.org/web/20221201134905/https://gsv.aero/schedule/", "https://web.archive.org/web/20230131103516/https://gsv.aero/schedule/", "https://web.archive.org/web/20230316002601/https://gsv.aero/schedule/", "https://web.archive.org/web/20230321112404/https://gsv.aero/schedule/", "https://web.archive.org/web/20230325094918/https://gsv.aero/schedule/", "https://web.archive.org/web/20230522214114/https://gsv.aero/schedule/", "https://web.archive.org/web/20230922115310/https://gsv.aero/schedule/", "https://web.archive.org/web/20240413130909/https://gsv.aero/schedule/", "https://web.archive.org/web/20240621014644/https://gsv.aero/schedule/", "https://web.archive.org/web/20240621014644/https://gsv.aero/schedule/", "https://web.archive.org/web/20241003234610/https://gsv.aero/schedule/", "https://web.archive.org/web/20241012223320/https://gsv.aero/schedule/", "https://web.archive.org/web/20241203210726/https://gsv.aero/schedule/", "https://web.archive.org/web/20241213223720/https://gsv.aero/schedule/", "https://web.archive.org/web/20250115103050/https://gsv.aero/schedule/", "https://gsv.aero/schedule/" ]


# Собираем все данные
all_flights = []
successful_dates = []

for url in urls:
    print(f"\nОбработка: {url}")
    capture_date, flights = parse_schedule(url)

    if capture_date and flights:
        all_flights.extend(flights)
        successful_dates.append(capture_date)
        print(f"Успешно обработано {len(flights)} рейсов за {capture_date}")
    else:
        print(f"Не удалось обработать URL: {url}")

    time.sleep(10)  # Увеличиваем задержку между запросами

# Фильтруем None значения перед анализом
valid_flights = [f for f in all_flights if f['capture_date'] is not None]

# Сохраняем полные данные
if valid_flights:
    full_data = {
        "metadata": {
            "total_flights": len(valid_flights),
            "unique_destinations": len({f['destination'] for f in valid_flights}),
            "unique_airlines": len({f['airline'] for f in valid_flights}),
            "time_period": f"{min(f['capture_date'] for f in valid_flights)} to {max(f['capture_date'] for f in valid_flights)}",
            "successful_captures": len(successful_dates),
            "failed_captures": len(urls) - len(successful_dates)
        },
        "flights": valid_flights
    }

    # Сохранение в файл
    with open('full_schedule_data.json', 'w', encoding='utf-8') as f:
        json.dump(full_data, f, ensure_ascii=False, indent=4)

    print(f"\nСохранено {len(valid_flights)} рейсов в файл 'full_schedule_data.json'")
    print(f"Успешно обработано дат: {len(successful_dates)} из {len(urls)}")
else:
    print("Не удалось получить ни одного рейса. Проверьте подключение к интернету.")
```

```python
import json
import matplotlib.pyplot as plt
import pandas as pd
from collections import defaultdict
from tabulate import tabulate

# Словарь для замены кодов аэропортов на названия
airport_names = {
    'VKO': 'Внуково',
    'SGC': 'Сургут',
    'SVO': 'Шереметьево',
    'DME': 'Домодедово',
    'AYT': 'Анталия',
    'AER': 'Сочи (Адлер)',
    'SIP': 'Симферополь',
    'LED': 'Пулково',
    'SVX': 'Кольцово (Екатеринбург)',
    'KRR': 'Пашковский (Краснодар)',
    'GOI': 'Даболим (Гоа, Индия)',
    'MRV': 'Минеральные Воды',
    'ROV': 'Платов (Ростов-на-Дону)',
    'KZN': 'Казань',
    'AAQ': 'Анапа',
    'OVB': 'Толмачёво (Новосибирск)',
    'UFA': 'Уфа',
    'TJM': 'Рощино (Тюмень)',
    'EVN': 'Звартноц (Ереван, Армения)',
    'PFO': 'Пафос (Кипр)',
    'GYD': 'Гейдар Алиев (Баку, Азербайджан)',
    'SKG': 'Македония (Салоники, Греция)',
    'FEG': 'Фергана (Узбекистан)',
    'NUX': 'Новый Уренгой',
    'NBE': 'Энфида (Тунис)',
    'FRU': 'Манас (Бишкек, Кыргызстан)',
    'SSH': 'Шарм-эль-Шейх (Египет)',
    'MCX': 'Уйташ (Махачкала)',
    'DYU': 'Душанбе (Таджикистан)',
    'KGD': 'Храброво (Калининград)',
    'LBD': 'Худжанд (Таджикистан)',
    'KUF': 'Курумоч (Самара)',
    'HRG': 'Хургада (Египет)',
    'GRV': 'Грозный',
    'ZIA': 'Жуковский',
    'GOJ': 'Нижний Новгород',
    'KQT': 'Бохтар (Таджикистан)',
    'TJU': 'Куляб (Таджикистан)',
    'OSS': 'Ош (Кыргызстан)'
}

def replace_airport_codes(destination):
    # Разбиваем строку на части (на случай составных направлений)
    parts = destination.split()
    replaced = []
    for part in parts:
        # Проверяем, является ли часть кодом аэропорта (3 большие буквы)
        if len(part) == 3 and part.isalpha() and part.isupper():
            replaced.append(airport_names.get(part, part))
        else:
            replaced.append(part)
    return ' '.join(replaced)

# Загрузка данных из файла
with open('full_schedule_data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Преобразование в DataFrame и замена кодов на названия
flights = pd.DataFrame(data['flights'])
flights['destination'] = flights['destination'].apply(replace_airport_codes)

# Преобразование даты в datetime
flights['capture_date'] = pd.to_datetime(flights['capture_date'])

# Разделение данных на два периода
pre_2022 = flights[flights['capture_date'].dt.year <= 2022]
post_2022 = flights[flights['capture_date'].dt.year > 2022]

# Функция для подсчета направлений
def count_destinations(flights_df):
    destinations = defaultdict(int)
    for dest in flights_df['destination']:
        destinations[dest] += 1
    return dict(sorted(destinations.items(), key=lambda item: item[1], reverse=True))

# Получаем данные для каждого периода
pre_2022_dest = count_destinations(pre_2022)
post_2022_dest = count_destinations(post_2022)

# Находим изменения в направлениях
all_destinations = set(pre_2022_dest.keys()).union(set(post_2022_dest.keys()))
new_destinations = set(post_2022_dest.keys()) - set(pre_2022_dest.keys())
discontinued_destinations = set(pre_2022_dest.keys()) - set(post_2022_dest.keys())

# Создаем таблицы изменений
new_dest_table = [(dest, post_2022_dest[dest]) for dest in new_destinations]
discontinued_dest_table = [(dest, pre_2022_dest[dest]) for dest in discontinued_destinations]

print("НОВЫЕ НАПРАВЛЕНИЯ (появились после 2022 года):")
print(tabulate(sorted(new_dest_table, key=lambda x: -x[1]),
               headers=["Направление", "Кол-во рейсов"],
               tablefmt="pretty"))

print("\nИСЧЕЗНУВШИЕ НАПРАВЛЕНИЯ (были до 2022, но исчезли после):")
print(tabulate(sorted(discontinued_dest_table, key=lambda x: -x[1]),
               headers=["Направление", "Кол-во рейсов"],
               tablefmt="pretty"))

# Создаем комбинированный график для сравнения направлений
common_destinations = set(pre_2022_dest.keys()).intersection(set(post_2022_dest.keys()))
sorted_common = sorted(common_destinations,
                      key=lambda x: pre_2022_dest[x] + post_2022_dest[x],
                      reverse=True)

# Берем топ-15 направлений для наглядности
top_n = 15
sorted_common = sorted_common[:top_n]

pre_counts = [pre_2022_dest[d] for d in sorted_common]
post_counts = [post_2022_dest[d] for d in sorted_common]
labels = sorted_common

x = range(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(15, 8))
rects1 = ax.bar([i - width/2 for i in x], pre_counts, width,
                label='До и включая 2022', color='#1f77b4')
rects2 = ax.bar([i + width/2 for i in x], post_counts, width,
                label='После 2022', color='#ff7f0e')

# Добавляем текст и оформление
ax.set_title('Сравнение количества рейсов по направлениям до и после 2022 года',
             pad=20, fontsize=14)
ax.set_xlabel('Направления', labelpad=10)
ax.set_ylabel('Количество рейсов', labelpad=10)
ax.set_xticks(x)
ax.set_xticklabels(labels, rotation=45, ha='right')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Добавляем подписи значений
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate('{}'.format(height),
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=8)

autolabel(rects1)
autolabel(rects2)

fig.tight_layout()
plt.show()

# Дополнительный график: изменения в количестве рейсов
changes = []
for dest in common_destinations:
    change = post_2022_dest[dest] - pre_2022_dest[dest]
    changes.append((dest, change, pre_2022_dest[dest], post_2022_dest[dest]))

# Сортируем по величине изменения
changes_sorted = sorted(changes, key=lambda x: abs(x[1]), reverse=True)[:15]

# Разделяем на увеличение и уменьшение
increased = [x for x in changes_sorted if x[1] > 0]
decreased = [x for x in changes_sorted if x[1] < 0]

# Создаем график изменений
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Наибольшие изменения в количестве рейсов по направлениям', y=1.02)

# График увеличения
if increased:
    ax1.barh([x[0] for x in increased], [x[1] for x in increased], color='green')
    ax1.set_title('Наибольший прирост рейсов')
    ax1.set_xlabel('Увеличение количества рейсов')
    ax1.grid(axis='x', linestyle='--', alpha=0.7)

# График уменьшения
if decreased:
    ax2.barh([x[0] for x in decreased], [abs(x[1]) for x in decreased], color='red')
    ax2.set_title('Наибольшее сокращение рейсов')
    ax2.set_xlabel('Уменьшение количества рейсов')
    ax2.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import pandas as pd
from collections import defaultdict
from tabulate import tabulate

# Список российских авиакомпаний (обновленный)
russian_airlines = {
    'Аэрофлот', 'S7 Airlines', 'Уральские авиалинии', 'Победа', 'Россия',
    'Nordwind Airlines', 'Azur Air', 'Red Wings', 'Ямал', 'ИрАэро',
    'Smartavia', 'Азимут', 'ЮВТ Аэро', 'Алроса', 'Ижавиа', 'Газпромавиа',
    'UVT AERO', 'РусЛайн', 'ЮТэйр', 'Пегас Флай', 'Royal Flight', 'NORD WIND', 'Pegas fly', 'ИКАР'
}

# Загрузка данных
with open('full_schedule_data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

flights = pd.DataFrame(data['flights'])
flights['capture_date'] = pd.to_datetime(flights['capture_date'])
flights['year'] = flights['capture_date'].dt.year

# Функция для определения типа авиакомпании
def classify_airline(airline):
    return 'Российская' if any(russian in airline for russian in russian_airlines) else 'Иностранная'

flights['airline_type'] = flights['airline'].apply(classify_airline)

# Разделяем данные на два периода: до 2022 (включительно) и после
pre_2022 = flights[flights['year'] <= 2022]
post_2022 = flights[flights['year'] > 2022]

# Получаем списки авиакомпаний для каждого периода
airlines_pre = set(pre_2022['airline'].unique())
airlines_post = set(post_2022['airline'].unique())

# Анализ изменений относительно 2022 года
new_after_2022 = airlines_post - airlines_pre
disappeared_after_2022 = airlines_pre - airlines_post
common_airlines = airlines_pre & airlines_post

# Создаем таблицы изменений
new_airlines_table = []
for airline in new_after_2022:
    first_year = post_2022[post_2022['airline'] == airline]['year'].min()
    count = len(post_2022[post_2022['airline'] == airline])
    new_airlines_table.append({
        'Авиакомпания': airline,
        'Год появления': first_year,
        'Кол-во рейсов': count,
        'Тип': classify_airline(airline)
    })

disappeared_airlines_table = []
for airline in disappeared_after_2022:
    last_year = pre_2022[pre_2022['airline'] == airline]['year'].max()
    count = len(pre_2022[pre_2022['airline'] == airline])
    disappeared_airlines_table.append({
        'Авиакомпания': airline,
        'Год исчезновения': last_year,
        'Кол-во рейсов': count,
        'Тип': classify_airline(airline)
    })

# Сортируем таблицы
new_airlines_table = sorted(new_airlines_table, key=lambda x: x['Год появления'])
disappeared_airlines_table = sorted(disappeared_airlines_table, key=lambda x: -x['Кол-во рейсов'])

# Выводим таблицы изменений
print("\nНОВЫЕ АВИАКОМПАНИИ (появились после 2022 года):")
print(tabulate([(x['Авиакомпания'], x['Год появления'], x['Кол-во рейсов'], x['Тип'])
                for x in new_airlines_table],
               headers=['Авиакомпания', 'Год появления', 'Кол-во рейсов', 'Тип'],
               tablefmt='pretty'))

print("\nИСЧЕЗНУВШИЕ АВИАКОМПАНИИ (были до 2022, но исчезли после):")
print(tabulate([(x['Авиакомпания'], x['Год исчезновения'], x['Кол-во рейсов'], x['Тип'])
                for x in disappeared_airlines_table],
               headers=['Авиакомпания', 'Последний год', 'Кол-во рейсов', 'Тип'],
               tablefmt='pretty'))

# График: сравнение количества авиакомпаний до и после 2022
plt.figure(figsize=(10, 6))
categories = ['До 2022 (вкл.)', 'После 2022']
russian_counts = [
    pre_2022[pre_2022['airline_type'] == 'Российская']['airline'].nunique(),
    post_2022[post_2022['airline_type'] == 'Российская']['airline'].nunique()
]
foreign_counts = [
    pre_2022[pre_2022['airline_type'] == 'Иностранная']['airline'].nunique(),
    post_2022[post_2022['airline_type'] == 'Иностранная']['airline'].nunique()
]

x = range(len(categories))
width = 0.35

plt.bar(x, russian_counts, width, label='Российские', color='#1f77b4')
plt.bar([i + width for i in x], foreign_counts, width, label='Иностранные', color='#ff7f0e')

plt.title('Количество авиакомпаний до и после 2022 года')
plt.xlabel('Период')
plt.ylabel('Количество авиакомпаний')
plt.xticks([i + width/2 for i in x], categories)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# График: изменение доли рейсов российских/иностранных авиакомпаний
pre_counts = pre_2022['airline_type'].value_counts()
post_counts = post_2022['airline_type'].value_counts()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
fig.suptitle('Доля рейсов российских и иностранных авиакомпаний')

ax1.pie(pre_counts, labels=pre_counts.index, autopct='%1.1f%%',
        colors=['#1f77b4', '#ff7f0e'], startangle=90)
ax1.set_title('До 2022 года (вкл.)')

ax2.pie(post_counts, labels=post_counts.index, autopct='%1.1f%%',
        colors=['#1f77b4', '#ff7f0e'], startangle=90)
ax2.set_title('После 2022 года')

plt.tight_layout()
plt.show()

# Динамика топ-5 авиакомпаний в каждом периоде
def get_top_airlines(df, n=5):
    return df['airline'].value_counts().head(n).index.tolist()

top_pre = get_top_airlines(pre_2022)
top_post = get_top_airlines(post_2022)

# Создаем DataFrame для сравнения
comparison_data = []
for airline in set(top_pre + top_post):
    pre_count = len(pre_2022[pre_2022['airline'] == airline])
    post_count = len(post_2022[post_2022['airline'] == airline]) if airline in airlines_post else 0
    comparison_data.append({
        'Авиакомпания': airline,
        'До 2022': pre_count,
        'После 2022': post_count,
        'Изменение': post_count - pre_count,
        'Тип': classify_airline(airline)
    })

comparison_df = pd.DataFrame(comparison_data).sort_values('До 2022', ascending=False)

# Визуализация сравнения топ-авиакомпаний
plt.figure(figsize=(12, 6))
x = range(len(comparison_df))
width = 0.35

colors = ['#1f77b4' if t == 'Российская' else '#ff7f0e' for t in comparison_df['Тип']]

plt.bar(x, comparison_df['До 2022'], width, label='До 2022', color='#1f77b4')
plt.bar([i + width for i in x], comparison_df['После 2022'], width, label='После 2022', color='#ff7f0e')

plt.title('Сравнение активности топ-авиакомпаний до и после 2022 года')
plt.xlabel('Авиакомпании')
plt.ylabel('Количество рейсов')
plt.xticks([i + width/2 for i in x], comparison_df['Авиакомпания'], rotation=45, ha='right')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import time
import requests
from bs4 import BeautifulSoup
import re
import json

# Список дней недели для замены "ежедневно"
days_of_week = ['пн', 'вт', 'ср', 'чт', 'пт', 'сб', 'вс']

def parse_schedule(url):
    try:
        # Извлекаем дату из URL
        date_match = re.search(r'/web/(\d{4})(\d{2})(\d{2})', url)
        if date_match:
            date = f"{date_match.group(1)}-{date_match.group(2)}-{date_match.group(3)}"
        else:
            date = "N/A"  # Если дата не найдена в URL

        response = requests.get(url)
        response.raise_for_status()  # Проверка на ошибки HTTP
        soup = BeautifulSoup(response.text, 'html.parser')

        table = soup.find('table')
        if not table:
            print(f"Таблица не найдена на странице: {url}")
            return []

        rows = table.find_all('tr')
        flights = []
        for row in rows[1:]:
            cols = row.find_all('td')
            if len(cols) == 8:
                days_flight = cols[2].text.strip().lower()

                # Если "ежедневно" — заменяем на список всех дней
                if "ежедневно" in days_flight:
                    days_flight = days_of_week
                # Иначе оставляем как есть (может быть "пн, ср, пт" и т.д.)
                else:
                    days_flight = [day.strip() for day in days_flight.split(',')]

                flight = {
                    "Дата расписания": date,
                    "Отправление": cols[0].text.strip(),
                    "Прибытие": cols[1].text.strip(),
                    "Дни полетов": days_flight,  # Теперь это список
                    "Период": cols[3].text.strip(),
                    "Авиакомпания": cols[4].text.strip(),
                    "Номер рейса": cols[5].text.strip(),
                    "Тип судна": cols[6].text.strip(),
                    "Аэропорт": cols[7].text.strip()
                }
                flights.append(flight)
        return flights
    except requests.exceptions.RequestException as e:
        print(f"Ошибка при запросе к {url}: {e}")
        return []

# Список URL-адресов для парсинга
urls = [ "https://web.archive.org/web/20190816013535/https://gsv.aero/schedule/", "https://web.archive.org/web/20190819090645/https://gsv.aero/schedule/", "https://web.archive.org/web/20190827183142/https://gsv.aero/schedule/", "https://web.archive.org/web/20200716105639/https://gsv.aero/schedule/", "https://web.archive.org/web/20200813172321/https://gsv.aero/schedule/", "https://web.archive.org/web/20200907230457/https://gsv.aero/schedule/", "https://web.archive.org/web/20200926021205/https://gsv.aero/schedule/", "https://web.archive.org/web/20201130142132/https://gsv.aero/schedule/", "https://web.archive.org/web/20210123013501/https://gsv.aero/schedule/", "https://web.archive.org/web/20210130204528/https://gsv.aero/schedule/", "https://web.archive.org/web/20210324080635/https://gsv.aero/schedule/", "https://web.archive.org/web/20210325051725/https://gsv.aero/schedule/", "https://web.archive.org/web/20210414140914/https://gsv.aero/schedule/", "https://web.archive.org/web/20210621041555/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210621041555/https://gsv.aero/schedule/", "https://web.archive.org/web/20210622085134/https://gsv.aero/schedule/", "https://web.archive.org/web/20210801233520/https://gsv.aero/schedule/", "https://web.archive.org/web/20210909135810/https://gsv.aero/schedule/", "https://web.archive.org/web/20210929020818/https://gsv.aero/schedule/", "https://web.archive.org/web/20211028175255/https://gsv.aero/schedule/", "https://web.archive.org/web/20220123060951/https://gsv.aero/schedule/", "https://web.archive.org/web/20220128073221/https://gsv.aero/schedule/", "https://web.archive.org/web/20220319023258/https://gsv.aero/schedule/", "https://web.archive.org/web/20220515100822/https://gsv.aero/schedule/", "https://web.archive.org/web/20220526210722/https://gsv.aero/schedule/", "https://web.archive.org/web/20221006204547/https://gsv.aero/schedule/", "https://web.archive.org/web/20221201134905/https://gsv.aero/schedule/", "https://web.archive.org/web/20230131103516/https://gsv.aero/schedule/", "https://web.archive.org/web/20230316002601/https://gsv.aero/schedule/", "https://web.archive.org/web/20230321112404/https://gsv.aero/schedule/", "https://web.archive.org/web/20230325094918/https://gsv.aero/schedule/", "https://web.archive.org/web/20230522214114/https://gsv.aero/schedule/", "https://web.archive.org/web/20230922115310/https://gsv.aero/schedule/", "https://web.archive.org/web/20240413130909/https://gsv.aero/schedule/", "https://web.archive.org/web/20240621014644/https://gsv.aero/schedule/", "https://web.archive.org/web/20240621014644/https://gsv.aero/schedule/", "https://web.archive.org/web/20241003234610/https://gsv.aero/schedule/", "https://web.archive.org/web/20241012223320/https://gsv.aero/schedule/", "https://web.archive.org/web/20241203210726/https://gsv.aero/schedule/", "https://web.archive.org/web/20241213223720/https://gsv.aero/schedule/", "https://web.archive.org/web/20250115103050/https://gsv.aero/schedule/", "https://gsv.aero/schedule/" ]

# Парсим данные для каждой ссылки
all_flights = []
for url in urls:
    print(f"Парсинг данных с: {url}")
    flights = parse_schedule(url)
    all_flights.extend(flights)
    time.sleep(10)  # Задержка между запросами

# Сохранение данных в файл
with open('all_flights.json', 'w', encoding='utf-8') as f:
    json.dump(all_flights, f, ensure_ascii=False, indent=4)

# Скачивание файла на ПК
from google.colab import files
files.download('all_flights.json')

print("Данные сохранены в файл 'all_flights.json' и готовы к скачиванию")
```
